# Wide vs Long Data in Pandas

```python
import pandas as pd
```

# -------------------------------
# 1. Create a Simple Wide Dataset
# -------------------------------



```python
# Example: Team members tracking how many times they ate out in two weeks
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Week1": [2, 3, 1],
    "Week2": [4, 2, 3],
    "Week3": [1, 2, 3],
    "Week4": [4, 2, 3]
}

df_wide = pd.DataFrame(data)
print("Wide Format Data:")
df_wide
```

# -------------------------------
# 2. Convert Wide → Long Format
# -------------------------------

```python
df_long = df_wide.melt(id_vars=["Name"],
                       var_name="Week",
                       value_name="Times_Ate_Out")

print("\nLong Format Data:")
df_long
```

# -------------------------------
# 3. Why Long is Useful
# -------------------------------

```python
# Example: Average times ate out across all people by week 
# 横着的可能会有不同的data type
avg_by_week = df_long.groupby("Week")["Times_Ate_Out"].mean().reset_index()

print("\nAverage Times Ate Out by Week (from long format):")
avg_by_week
```

```python
# Example: Average times ate out across all week by people
avg_by_name= df_long.groupby("Name")["Times_Ate_Out"].mean().reset_index()

print("\nAverage Times Ate Out by Name (from long format):")
avg_by_name
```

# -------------------------------
# 4. Visualize (Optional)
# -------------------------------

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.barplot(x = "Week", y = "Times_Ate_Out", data = avg_by_week)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.barplot(x = "Name", y = "Times_Ate_Out", data = avg_by_name)

plt.show()
```
